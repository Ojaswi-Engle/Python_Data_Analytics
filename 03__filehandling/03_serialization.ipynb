#serialization and deserialzation
serialization - process of converting python object to JSON format
1. python object to json string
2.python object to json file(write)

```python
import json
py={
  'name':'ojaswi',
  'age':20,
  'Branch':'CSE'
}
s=json.dumps(py)
print(s)
print(type(s))
```

```python
import json
py={
  'name':'ojaswi',
  'age':20,
  'Branch':'CSE'
}
with open('sample.json','w') as f:
  json.dump(py,f)
```

deserialization - process of converting Json format to python object
1.Json string to py obj
2.json file to py object


```python
import json
js='{"name": "ojaswi", "age": 20, "Branch": "CSE"}'
py=json.loads(js)
print(py)
print(type(py))
```

```python
import json
with open('sample.json','r') as f:
  data=json.load(f)
  print(data)
  print(type(data))
```

```python
#serialisation and deserialization of tuples
import json
t=(1,2,3,4,5,6)

with open('t.json','w') as f:
  json.dump(t,f)

import json
with open('t.json','r') as f:
  data=json.load(f)
  print(data)
  print(type(data))
```

```python
import json
class Person:
  def __init__(self,fname,lname,age,gender):
    self.fname=fname
    self.lname=lname
    self.age=age
    self.gender=gender

p=Person('Ojaswi','Engle',21,'female')

def convert_obj(p):
  return {'first_Name':p.fname,
          'last_name':p.lname,
          'age':p.age,
          'gender':p.gender
          }

with open('obj.json','w') as f:
  json.dump(p,f,default=convert_obj,indent=3)

import  json
with open('obj.json','r') as f:
  data=json.load(f)  #read data is not custom object rather it is dictionary original object is lost
  print(data)
  print(type(data))

```
