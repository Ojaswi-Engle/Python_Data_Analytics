```python
#pickling
import pickle

class Student:
  def __init__(self,name,age,branch,marks):
    self.name=name
    self.age=age
    self.branch=branch
    self.marks=marks
s = Student('Ojaswi',21,'CSE',[78,89,90,67,79])

with open('obj.pkl','wb') as f:
  pickle.dump(s,f)



```

```python
import pickle
with open('obj.pkl','rb') as f:
  data=pickle.load(f)
  print(data)
  print(type(data))
  print(data.name)
  print(data.age)
  print(data.branch)
  print(data.marks)
```
